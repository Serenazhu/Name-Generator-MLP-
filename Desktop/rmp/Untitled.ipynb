```python
import pandas as pd
path1 = "/Users/serenazhu/Desktop/rmp/FA_2024.csv"
path2 = "/Users/serenazhu/Desktop/rmp/SU_2024.csv"

data1 = pd.read_csv(path1)
data2 = pd.read_csv(path2)

all_df = pd.concat([data1, data2], ignore_index=True)
sum_cols = ["A","B","C","D","F","Total"]
rate_cols = ["Compl","Succ."]

merged = (
    all_df.groupby("Name", as_index=False)
    .agg({
        "Dept": "first", 
        **{c: "sum" for c in sum_cols},
        **{c: "mean" for c in rate_cols},}
    )
)
merged.head()
```
